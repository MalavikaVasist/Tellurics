# Testing Resolution: Three telluric generation methods

Load the convolved Phoenix FITS (R=100k), then generate one telluric
transmission (0.8–0.95 µm) three different ways:

- **(a)** Native LBLRTM (no wavegrid, no resolution) → convolve to R=100k → resample onto Phoenix grid
- **(b)** `MakeModel(resolution=100000)` — TelFit's internal convolution
- **(c)** `MakeModel(wavegrid=phoenix_grid)` — rebin only, no convolution

Time each, then compare.

```python
import sys
import time
import warnings
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt
from astropy.io import fits
from scipy.ndimage import gaussian_filter1d

# warnings.filterwarnings("ignore", message="pkg_resources is deprecated.*")
# warnings.filterwarnings("ignore", message="PYSYN_CDBS is undefined.*")
# warnings.filterwarnings("ignore", message="Extinction files not found.*")

TELFIT_SRC = Path("/home/mvasist/Documents/Tellurics/TelFit/src")
sys.path.insert(0, str(TELFIT_SRC))
from telfit import Modeler
```

## 1. Configuration

```python
PHOENIX_DIR = Path("/home/mvasist/Documents/Tellurics/Phoenix")
WORKDIR     = Path.home() / ".TelFit" / "rundir1"

# Convolved Phoenix FITS produced by testing_stellar_spec.ipynb
PHOENIX_FITS = PHOENIX_DIR / "high_res" / "phoenix_A0V_T9600_logg4.0_feh+0.0_R100000_s3_convolved.fits"

WAVESTART_NM = 800.0
WAVEEND_NM   = 950.0
R_TARGET     = 100_000
WAVE_PADDING_NM = 0.5
FWHM_TO_SIGMA = 2.0 * np.sqrt(2.0 * np.log(2.0))

TELFIT_PARAMS = dict(
    pressure=790.0, temperature=285.0, humidity=50.0, angle=45.0,
    lat=30.6, alt=2.1, co2=368.5, o3=3.9e-2, n2o=1e-4, co=0.14,
    ch4=1.8, o2=2.1e5, no=1.1e-19, so2=1e-4, no2=1e-4, nh3=1e-4, hno3=5.6e-4,
)
```

## 2. Load the Phoenix R=100k grid

```python
with fits.open(PHOENIX_FITS) as hdul:
    phoenix_data = hdul['SPECTRUM'].data

phoenix_wave_nm = np.asarray(phoenix_data['WAVE_NANO'], dtype=np.float64) 
phoenix_flux = np.asarray(phoenix_data['FLUX'], dtype=np.float64) 

print(f"Phoenix grid: {len(phoenix_wave_nm)} points")
print(f"Range: {phoenix_wave_nm[0]:.4f} - {phoenix_wave_nm[-1]:.4f} nm")
```

## 3. Helper function

```python
def convolve_to_R_and_resample(wave_nm, transmission, target_wave_nm, target_R):
    """Convolve a native TelFit spectrum to target_R, resample onto target_wave_nm.
    Take every tiny piece of light in my theoretical spectrum, spread it according to
    the telescope's Gaussian response, and add all the overlapping contributions together
    """
    w = np.asarray(wave_nm, dtype=float)
    t = np.asarray(transmission, dtype=float)

    log_w = np.log(w)
    dlog = float(np.median(np.diff(log_w)))
    log_uniform = np.arange(log_w[0], log_w[-1] + 0.5 * dlog, dlog)
    t_uniform = np.interp(log_uniform, log_w, t)

    sigma_pixels = 1.0 / (target_R * FWHM_TO_SIGMA) / dlog
    t_conv = gaussian_filter1d(t_uniform, sigma=sigma_pixels, mode='nearest', truncate=4.0)

    return np.interp(np.log(target_wave_nm), log_uniform, t_conv)
```

```python
modeler = Modeler(print_lblrtm_output=False, debug=False)
print("Modeler ready")
```

## 4. Method (a): Native TelFit -> convolve to R=100k -> resample onto Phoenix grid

```python
lowfreq_a  = 1e7 / (WAVEEND_NM + WAVE_PADDING_NM)
highfreq_a = 1e7 / (WAVESTART_NM - WAVE_PADDING_NM)

t0 = time.time()
model_a = modeler.MakeModel(
    lowfreq=lowfreq_a, highfreq=highfreq_a,
    workdir=str(WORKDIR), **TELFIT_PARAMS,
)

plt.plot(model_a.x, model_a.y)

trans_a = convolve_to_R_and_resample(
    np.array(model_a.x), np.array(model_a.y), phoenix_wave_nm, R_TARGET,
)
time_a = time.time() - t0

print(f"(a) Native -> convolve: {time_a:.1f} s")
print(f"    Native points: {len(model_a.x)}, output: {len(trans_a)}")
```

### Plotting the Pheonix grid with the telluric

```python
plt.plot(phoenix_wave_nm, trans_a/10e-16, label = 'Telluric')
plt.plot(phoenix_wave_nm, phoenix_flux, label = 'Phenix spectrum')
plt.show()
```

## 5. Method (b): TelFit with resolution=100000 and interpolate to Pheonix

```python
t0 = time.time()
model_b = modeler.MakeModel(
    lowfreq=1e7 / WAVEEND_NM, highfreq=1e7 / WAVESTART_NM,
    resolution=float(R_TARGET),
    workdir=str(WORKDIR), **TELFIT_PARAMS,
)
wave_b = np.array(model_b.x, dtype=np.float64)
trans_b_raw = np.array(model_b.y, dtype=np.float64)
time_b = time.time() - t0

# TelFit returns on its own grid, so interpolate onto Phoenix grid
trans_b = np.interp(phoenix_wave_nm, wave_b, trans_b_raw)

print(f"(b) resolution=100k: {time_b:.1f} s")
print(f"    TelFit returned: {len(wave_b)} points")
```

## 6. Method (c): TelFit with just interpolating at wavegrid=phoenix_grid

```python
t0 = time.time()
model_c = modeler.MakeModel(
    lowfreq=1e7 / WAVEEND_NM, highfreq=1e7 / WAVESTART_NM,
    wavegrid=phoenix_wave_nm.astype(np.float32),
    workdir=str(WORKDIR), **TELFIT_PARAMS,
)
wave_c = np.array(model_c.x, dtype=np.float64)
trans_c = np.array(model_c.y, dtype=np.float64)
time_c = time.time() - t0

# Already on Phoenix grid - no interpolation needed
print(f"(c) wavegrid=Phoenix: {time_c:.1f} s")
print(f"    Returned: {len(wave_c)} points")
print(f"    Grid matches Phoenix: {np.allclose(wave_c, phoenix_wave_nm)}")
```

## 7. Timing summary

```python
print(f"{'Method':<50} {'Time (s)':>10}")
print('-' * 62)
print(f"{'(a) Native -> convolve to R=100k':<50} {time_a:>10.1f}")
print(f"{'(b) resolution=100k':<50} {time_b:>10.1f}")
print(f"{'(c) wavegrid=Phoenix':<50} {time_c:>10.1f}")
```

## 8. Compare: overlay all three

```python
res_ab = trans_a - trans_b
res_ac = trans_a - trans_c
res_bc = trans_b - trans_c

fig, axes = plt.subplots(5, 1, figsize=(14, 14), sharey = True)

# Panel 1: all three
ax = axes[0]
ax.plot(phoenix_wave_nm, trans_a, lw=0.5, alpha=0.8, label='(a) native -> convolve')
ax.plot(phoenix_wave_nm, trans_b, lw=0.5, alpha=0.8, label='(b) resolution=100k')
ax.plot(phoenix_wave_nm, trans_c, lw=0.5, alpha=0.8, label='(c) wavegrid=Phoenix')
ax.set_ylabel('Transmission')
ax.set_title('All three methods (800-950 nm)')
ax.legend(fontsize=9)
ax.set_ylim(-0.05, 1.05)

# Panel 2: residual (a) - (b)
ax = axes[1]
ax.plot(phoenix_wave_nm, res_ab, lw=0.4, color='C3')
ax.axhline(0, ls='--', color='k', lw=0.5)
ax.set_ylabel('(a) - (b)')
ax.set_title(f'Residual (a)-(b): std={np.std(res_ab):.2e}, max|res|={np.max(np.abs(res_ab)):.2e}')

# Panel 3: residual (a) - (c)
ax = axes[2]
ax.plot(phoenix_wave_nm, res_ac, lw=0.4, color='C4')
ax.axhline(0, ls='--', color='k', lw=0.5)
ax.set_ylabel('(a) - (c)')
ax.set_title(f'Residual (a)-(c): std={np.std(res_ac):.2e}, max|res|={np.max(np.abs(res_ac)):.2e}')

# Panel 2: residual (a) - (b)
ax = axes[3]
ax.plot(phoenix_wave_nm, res_bc, lw=0.4, color='C5')
ax.axhline(0, ls='--', color='k', lw=0.5)
ax.set_ylabel('(b) - (c)')
ax.set_title(f'Residual (b)-(c): std={np.std(res_bc):.2e}, max|res|={np.max(np.abs(res_bc)):.2e}') 

# Panel 4: zoom 900-920 nm
ax = axes[4]
zm = (phoenix_wave_nm >= 900) & (phoenix_wave_nm <= 901)
ax.plot(phoenix_wave_nm[zm], trans_a[zm], lw=1, label='(a) native -> convolve')
ax.plot(phoenix_wave_nm[zm], trans_b[zm], lw=1, ls='--', label='(b) resolution=100k')
ax.plot(phoenix_wave_nm[zm], trans_c[zm], lw=1, ls=':', label='(c) interpolating wavegrid=Phoenix')
ax.set_xlabel('Wavelength (nm)')
ax.set_ylabel('Transmission')
ax.set_title('Zoom: 900-920 nm (water band)')
ax.legend(fontsize=9)

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].hist(res_ab, bins=100, edgecolor='k', alpha=0.7, color='C3')
axes[0].axvline(0, ls='--', color='r')
axes[0].set_xlabel('(a) - (b)')
axes[0].set_title('Histogram: (a)-(b)')

axes[1].hist(res_ac, bins=100, edgecolor='k', alpha=0.7, color='C4')
axes[1].axvline(0, ls='--', color='r')
axes[1].set_xlabel('(a) - (c)')
axes[1].set_title('Histogram: (a)-(c)')

plt.tight_layout()
plt.show()
```

## Summary

| Method | How | Convolution? | Expected behavior |
|---|---|---|---|
| **(a)** native -> convolve | LBLRTM at R~millions -> Gaussian in ln(lam) -> resample | Yes | True R=100k (ground truth) |
| **(b)** resolution= | TelFit internal (script 17 says it does not work) | Unclear | Compare with (a) to find out |
| **(c)** wavegrid= | LBLRTM -> rebin/interpolate onto grid | No | Grid is R=100k but spectral content is higher |
